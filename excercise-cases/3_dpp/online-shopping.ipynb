 The dataset records information about web browsing sessions. The web shop uses a taxonomy in which pages are tagged depending on their main contet as:
- administrative
- informational

In the dataset, each row represents a browsing session. `Revenue` indicates whether the session resulted in a purchase. Other columns have the following meaning.

| Feature | Meaning |
|---|---|
| `Administrative` | Number of administrative pages visited during the session. |
| `Administrative_Duration` | Total time spent on administrative pages, in seconds. |
| `Informational` | Number of informational pages visited during the session. |
| `Informational_Duration` | Total time spent on informational pages, in seconds. |
| `ProductRelated` | Number of product-related pages visited during the session. |
| `ProductRelated_Duration` | Total time spent on product-related pages, in seconds. |
| `BounceRates` | Average bounce rate of the pages visited during the session. |
| `ExitRates` | Average exit rate of the pages visited during the session. |
| `PageValues` | Average page value of pages visited before the session ended. |
| `SpecialDay` | Closeness of the session date to a special day (such as a holiday), represented numerically. |
| `Month` | Month in which the session occurred. |
| `OperatingSystems` | Coded category for the visitor’s operating system. |
| `Browser` | Coded category for the visitor’s browser. |
| `Region` | Coded geographic region of the visitor. |
| `TrafficType` | Coded category for the source or type of traffic. |
| `VisitorType` | Visitor category, such as returning or new visitor. |
| `Weekend` | Whether the session occurred on a weekend (`True`/`False`). |
| `Revenue` | Whether the session resulted in a purchase (`True`/`False`); this is commonly used as a prediction target. |

Run the following code cell command to download the dataset. It will only download it once and then store it in the `data` folder.

```python
from io import BytesIO
from pathlib import Path
from urllib.request import Request, urlopen
from zipfile import ZipFile
import shutil
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import RocCurveDisplay, ConfusionMatrixDisplay, classification_report
from matplotlib import pyplot as plt
SEED = 67  # Set the random seed for reproducibility

url = (
    "https://archive.ics.uci.edu/static/public/468/"
    "online+shoppers+purchasing+intention+dataset.zip"
)
csv_name = "online_shoppers_intention.csv"
data_dir = Path("../../data")
csv_path = data_dir / csv_name

if csv_path.is_file():
    print(f"Loading existing dataset: {csv_path.resolve()}")
else:
    data_dir.mkdir(parents=True, exist_ok=True)

    request = Request(url, headers={"User-Agent": "Python dataset downloader"})
    with urlopen(request, timeout=60) as response:
        archive_data = response.read()

    with ZipFile(BytesIO(archive_data)) as archive:
        csv_member = next(
            (
                name for name in archive.namelist()
                if Path(name).name == csv_name
            ),
            None,
        )
        if csv_member is None:
            raise FileNotFoundError(f"{csv_name} was not found in the downloaded archive.")

        with archive.open(csv_member) as source, csv_path.open("wb") as destination:
            shutil.copyfileobj(source, destination)

    print(f"Downloaded dataset to: {csv_path.resolve()}")

df = pd.read_csv(csv_path)

print(f"Shape: {df.shape}")
display(df.head())
```

```python
# Check you did your homework
def homework_done():
    if "dru" not in globals():
        print("Hey dru, where are you? 👀")
        return False
    data = globals()["dru"]
    if not isinstance(data, pd.DataFrame):
        print("Hey dru, what's wrong with you? 🔍")
        return False
    if "Revenue" not in data.columns:
        print("No revenue, dru?")
        return False
    if data["Revenue"].nunique() != 2:
        print("Oh dru, I was expecting two distinct Revenue values 💰")
        return False
    if not all(
        pd.api.types.is_numeric_dtype(dtype)
        or pd.api.types.is_bool_dtype(dtype)
        for dtype in data.dtypes
    ):
        print("Sorry dru, can't let you pass. Sob, sob 😥")
        return False
    print("🏆 Well done! You are amazing 🤗")
    return True

def run_predictive_analytics_workflow(df):
    """ Implements a simple predictive analytics workflow using logistic regression on the provided DataFrame. 

        We assume that the DataFrame contains a binary target variable named 'Revenue' and various features for prediction. The function performs the following steps:
        1. Splits the data into training and testing sets.
        2. Fits a logistic regression model to the training data.
        3. Predicts the probabilities of the positive class for the test set.
        4. Assesses the model performance using ROC analysis, confusion matrix, and classification report

        No checking is done on the input DataFrame, so it is assumed to be preprocessed and ready for modeling.
        Any deviances from the expected structure of the DataFrame will cause errors during execution. 
        This is why you should always do your homework 😉
    
    """
    # Separate features and the target
    X = df.copy()
    y = X.pop("Revenue").astype(int)
    print(f"Shape of X: {X.shape}, Shape of y: {y.shape}") 

    # 1. Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=SEED)
    # 2. Fit the logistic regression model (we use L1 regularization, and explain another day what that is)
    lr = LogisticRegression(max_iter=1000, l1_ratio=1.0, solver="liblinear", random_state=SEED)
    lr.fit(X_train, y_train)
    # 3. Predict the probabilities of the positive class (1) for the test set
    yhat = lr.predict_proba(X_test)[:, 1]
    # 4. Assess the model performance using the ROC analysis, confusion matrix, and classification report
    tau = 0.5  # threshold for classification
    # ROC confusion matrix and ROC curve
    fig, (ax_cm, ax_roc) = plt.subplots(1, 2, figsize=(12, 5))
    ConfusionMatrixDisplay.from_predictions(
        y_test,
        yhat > tau,
        cmap="Blues",
        display_labels=["No Purchase", "Purchase"],
        ax=ax_cm,
    )

    RocCurveDisplay.from_predictions(
        y_test,
        yhat,
        plot_chance_level=True,
        ax=ax_roc,
    )
    plt.tight_layout()
    plt.show()
    # Classification report
    c_report = classification_report(y_test, yhat > tau, target_names=["No Purchase", "Purchase"], output_dict=False)
    print("\n\nCLASSIFICATION REPORT:\n")
    print(c_report)

if homework_done():
    run_predictive_analytics_workflow(dru)

```
