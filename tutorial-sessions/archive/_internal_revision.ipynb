## 5.1 Concept Recap
A key advantage of ensemble methods is the improvement in the bias-variance trade-off. Ensemble models can reduce bias or variance or both.  

Bias: How closely a model's average predictions match the true outcomes.
Variance: How much the model's predictions change when trained on different subsets of data.

Combining multiple models reduces variance because each model’s errors can (partly) cancel out when averaged. With well-chosen base models or through methods like boosting, we ensemble models can reduce bias as well.

## 5.2 Programming Demo: Visualizing Bias-Variance with multiple models
Here is an illustrative example. We fit several base learners and compare their accuracies and variance in results across multiple random splits of the data into training and test sets. 

```python
from sklearn.metrics import accuracy_score
# We'll create a function to estimate variance of accuracy across multiple runs
def estimate_variance_of_model(model, X, y, runs=10):
    scores = []
    for i in range(runs):
        X_train, X_test, y_train, y_test = train_test_split(X, y, 
                                                            test_size=0.2, 
                                                            random_state=i*10)
        model.fit(X_train, y_train)
        preds = model.predict(X_test)
        score = accuracy_score(y_test, preds)
        scores.append(score)
    return np.mean(scores), np.std(scores)

models = {
    'LogisticRegression': LogisticRegression(max_iter=1000),
    'RandomForest': RandomForestClassifier(n_estimators=30),
    'Bagging(LogisticRegression)': BaggingClassifier(
        base_estimator=LogisticRegression(max_iter=1000), 
        n_estimators=10
    ),
}

for name, m in models.items():
    mean_acc, std_acc = estimate_variance_of_model(m, X, y, runs=10)
    print(f"{name:30s} -> Mean Acc: {mean_acc:.3f}, Std: {std_acc:.3f}")
```
